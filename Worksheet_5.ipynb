```python
#Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#reading the csv file.
df = pd.read_csv('student.csv')

#Seperating X and y values.
X = df[['Math', 'Reading']].values
y = df['Writing'].values

d = X.shape[1] #Number of features.
W = np.zeros(d) #Initializing weights.

#Splitting the values for training and testing.
def training_testing_data(X, y, test_size=0.3, random_seed=42):
    np.random.seed(random_seed)
    n = X.shape[0]
    indices = np.random.permutation(n)
    test_count = int(test_size * n)
    test_idx = indices[:test_count]
    train_idx = indices[test_count:]

    X_train = X[train_idx]
    X_test = X[test_idx]
    y_train = y[train_idx]
    y_test = y[test_idx]

    return X_train, X_test, y_train, y_test

#Define the cost function Mean Squared Error
def cost_function(X, y, W):
    n = len(y)
    y_pred = X @ W
    error = y_pred - y
    cost = (1 / n) * np.sum(error ** 2)
    return cost

#implementing the Gradient Descent algorithm.
def gradient_descent(X, y, W, learning_rate=0.0001, epochs=1000):
    n = len(y)
    cost_history = []

    for i in range(epochs):
        y_pred = X @ W
        gradient = (2 / n) * (X.T @ (y_pred - y))
        W = W - learning_rate * gradient
        cost = cost_function(X, y, W)
        cost_history.append(cost)

    return W, cost_history

# Model Evaluation - RMSE
def rmse(Y, Y_pred):
    n = len(Y)
    rmse = np.sqrt((1 / n) * np.sum((Y_pred - Y) ** 2))
    return rmse

def main():
    data = pd.read_csv('student.csv')

    X = data[['Math', 'Reading']].values
    Y = data['Writing'].values

    X_train, X_test, Y_train, Y_test = training_testing_data(X, Y, test_size=0.2)

    W = np.zeros(X_train.shape[1])
    alpha = 0.0001
    iterations = 1000
    W_optimal, cost_history = gradient_descent(X_train, Y_train, W, alpha, iterations)
    Y_pred = np.dot(X_test, W_optimal)

    model_rmse = rmse(Y_test, Y_pred)

    print("Final Weights:", W_optimal)
    print("Cost History (First 10 iterations):", cost_history[:10])
    print("RMSE on Test Set:", model_rmse)

#Final step is calling the main file.
print(main())
```
