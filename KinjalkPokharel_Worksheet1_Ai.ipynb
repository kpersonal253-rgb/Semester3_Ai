## Problem 4 – Matrix Operations

```python
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])
A @ np.linalg.inv(A)
```

```python
A @ B, B @ A
```

```python
left = (A@B).T
right = B.T @ A.T
left, right
```
