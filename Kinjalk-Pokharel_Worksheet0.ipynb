## Recursion Task 1 - Sum of Nested List

```python
def sum_nested_list(x):
    total = 0
    for item in x:
        if isinstance(item, list):
            total += sum_nested_list(item)
        else:
            total += item
    return total

nested_list = [1, [2, [3,4],5], 6, [7,8]]
sum_nested_list(nested_list)
```

## Recursion Task 2 - String Permutations

```python
def generate_permutations(s):
    if len(s) <= 1:
        return [s]
    perms = []
    for i, ch in enumerate(s):
        for perm in generate_permutations(s[:i]+s[i+1:]):
            res = ch + perm
            if res not in perms:
                perms.append(res)
    return perms

generate_permutations("abc")
```

## Recursion Task 3 - Directory Size

```python
directory_structure = {'file1.txt': 200, 'file2.txt': 300, 'subdir1': {'file3.txt': 400, 'file4.txt': 100}, 'subdir2': {'subsubdir1': {'file5.txt': 250}, 'file6.txt': 150}}

def calculate_directory_size(d):
    total = 0
    for key, value in d.items():
        if isinstance(value, dict):
            total += calculate_directory_size(value)
        else:
            total += value
    return total

calculate_directory_size(directory_structure)
```

## DP Task 1 - Coin Change

```python
def min_coins(coins, amount):
    dp = [float('inf')] * (amount+1)
    dp[0] = 0
    for c in coins:
        for i in range(c, amount+1):
            dp[i] = min(dp[i], dp[i-c] + 1)
    return dp[amount] if dp[amount] != float('inf') else -1

min_coins([1,2,5], 11)
```

## DP Task 2 - Longest Common Subsequence

```python
def longest_common_subsequence(a,b):
    dp = [[0]*(len(b)+1) for _ in range(len(a)+1)]
    for i in range(1,len(a)+1):
        for j in range(1,len(b)+1):
            if a[i-1]==b[j-1]:
                dp[i][j]=dp[i-1][j-1]+1
            else:
                dp[i][j]=max(dp[i-1][j],dp[i][j-1])
    return dp[-1][-1]

longest_common_subsequence("abcde","ace")
```

## DP Task 3 - Knapsack

```python
def knapsack(weights, values, cap):
    n=len(weights)
    dp=[[0]*(cap+1) for _ in range(n+1)]
    for i in range(1,n+1):
        for w in range(1,cap+1):
            if weights[i-1] <= w:
                dp[i][w] = max(values[i-1] + dp[i-1][w-weights[i-1]], dp[i-1][w])
            else:
                dp[i][w] = dp[i-1][w]
    return dp[n][cap]

knapsack([1,3,4,5],[1,4,5,7],7)
```
